```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option(
    "display.max_columns",
    None
)

print("Libraries imported successfully.")

```

# STep 2:Load Model Results

## Purpose

Store evaluation metrics generated during model development.

## Description

The forecasting performance metrics from SARIMAX and XGBoost are collected for comparison.

These metrics represent each model's ability to predict future taxi demand.

## Expected Outcome

Performance metrics are available for comparison.

---

```python
sarimax_mae = 1543.93
sarimax_rmse = 1914.88
sarimax_r2 = 0.6455

xgb_mae = 6.04
xgb_rmse = 13.9
xgb_r2 = 0.9626
```

# Step 3:Create Model Comparison Table

## Purpose

Organize forecasting metrics into a single comparison framework.

## Description

A structured comparison table enables direct evaluation of forecasting performance across all developed models.

## Metrics

### MAE

Average prediction error.

Lower is better.

### RMSE

Overall forecasting error.

Lower is better.

### R²

Explained variance.

Higher is better.

## Expected Outcome

A consolidated model comparison table is created.

---

```python
comparison = pd.DataFrame({

    "Model": [
        "SARIMAX",
        "XGBoost"
    ],

    "MAE": [
        sarimax_mae,
        xgb_mae
    ],

    "RMSE": [
        sarimax_rmse,
        xgb_rmse
    ],

    "R2": [
        sarimax_r2,
        xgb_r2
    ]

})

comparison
```

# Step 4:Forecast Accuracy Comparison

## Purpose

Determine which model produces more accurate forecasts.

## Description

Forecasting accuracy is evaluated using MAE, RMSE, and R².

A model producing lower forecasting errors and higher explanatory power is considered superior.

## Expected Outcome

The most accurate forecasting model is identified.

---

```python
comparison.sort_values(
    "RMSE"
)
```

```python
comparison.set_index(
    "Model"
)[["MAE", "RMSE"]].plot(
    kind="bar",
    figsize=(10,5)
)

plt.title(
    "Forecast Accuracy Comparison"
)

plt.ylabel(
    "Error"
)

plt.show()
```
