# Model Deployment Using Streamlit

## Objective

The objective of this notebook is to prepare the final forecasting model for deployment and create an interactive application that allows users to generate taxi demand forecasts without retraining the model.

Deployment represents the final stage of the machine learning lifecycle, where a trained model is integrated into an application and made accessible to end users for inference.

---

## Why Deployment is Important

Developing a forecasting model is only one part of a machine learning project. To create business value, the model must be operationalized so that stakeholders can utilize it for decision-making.

Deployment enables:

- Real-time or on-demand demand forecasting.
- Integration of the model into business workflows.
- Accessibility for non-technical users.
- Automation of forecasting tasks.
- Reusability without retraining the model.

---

## Why Streamlit Was Selected

Several deployment options exist, including Flask, FastAPI, and cloud platforms. Streamlit was selected because it provides:

- Rapid development of interactive applications.
- Minimal backend development requirements.
- Easy integration with machine learning models.
- Simple deployment and demonstration capabilities.
- Strong suitability for academic projects and portfolio presentations.

## Step 1:Load Final Trained Model

### Why is this step necessary?

Training machine learning models can be computationally expensive. Saving and loading a trained model eliminates the need for retraining whenever predictions are required.

The final XGBoost model selected during model comparison was previously saved using Joblib.

Loading the saved model ensures:

- Consistency of forecasting results.
- Reduced computational overhead.
- Reproducibility of the deployment pipeline.

```python
import joblib

model = joblib.load(
    "/home/ed/Desktop/NYC_Taxi_Demand_Forecasting/Yellow_Dataset/models/final_xgboost_model.pkl"
)

print(
    "Final XGBoost model loaded successfully."
)
```

## Step 2:Save Feature Ordering Information

### Why is feature ordering important?

XGBoost expects prediction inputs to have the exact same feature names and ordering used during training.

If the deployment application provides features in a different order, prediction errors or incorrect forecasts may occur.

Saving the feature order ensures:

- Consistent model inputs.
- Prevention of feature mismatch errors.
- Reliable forecasting during deployment.

```python
import json

feature_order = model.get_booster().feature_names

with open(
    "/home/ed/Desktop/NYC_Taxi_Demand_Forecasting/Yellow_Dataset/models/feature_order.json",
    "w"
) as file:

    json.dump(
        feature_order,
        file
    )

print(
    "Feature ordering information saved successfully."
)
```
