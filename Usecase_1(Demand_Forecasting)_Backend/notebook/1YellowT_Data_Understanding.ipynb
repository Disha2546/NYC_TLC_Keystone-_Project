# Step 1: Import Required Libraries

We import the libraries required for data loading,
analysis and visualization.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Pandas display setting.
pd.set_option('display.max_columns',None)
```

# Step 2 : Load the Yellow Taxi dataset



```python
df = pd.read_parquet(
    "/home/ed/Desktop/NYC_Taxi_Demand_Forecasting/Yellow_Dataset/data/yellow_tripdata_2026-01.parquet"
)

print("Dataset loaded successfully.")
```

# Step 3: Dataset Shape

Check the number of rows and columns available in the dataset.

```python
print("Dataset shape:", df.shape)
```

# Step 4: Preview Dataset

Display the first few records to understand the dataset structure.

```python
print("First 5 rows of the dataset:")
print(df.head())
```

# Step 5: Dataset Information

Inspect column names, data types, non-null values, and memory usage.

```python
print("Dataset Information::")
df.info()
```

# Step 6: Column Names

Display all available columns in the dataset.

```python
print("Columns in Dataset:")
# print(df.columns.to_list())
for col in df.columns:
    print(col)
```

# Step 7: Missing Values

Check whether any columns contain missing values.

```python
missing = df.isnull().sum()
print("Missing values in  column:")
missing[missing > 0].sort_values(ascending=False)
```

# Step 8: Missing Value Percentage

Calculate the percentage of missing values in each column.

```python
print("Percentage of missing values in each column:")
(
    df.isnull()
      .mean()
      .mul(100)
      .round(2)
      .sort_values(ascending=False)
)
```

# Step 9: Duplicate Records Analysis

Check whether the dataset contains duplicate records.

Duplicate records can lead to incorrect analysis and may affect model performance if not handled properly.


```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates:,}")
```

# Step 10: Memory Usage

Measure the memory consumed by the dataset.

This helps estimate the computational resources required for analysis and model training.

```python
memory = (
    df.memory_usage(deep=True)
      .sum()
      /1024 / 1024
)

print(f"Memory Usage : {memory:.2f} MB")
```

# Step 11: Statistical Summary

Generate summary statistics for numerical columns.

This helps identify data distribution, extreme values, and potential outliers.

```python
print("Descriptive Statistics:")
print(df.describe())
```

# Step 12: Datetime Range Validation

Check the available date range in the dataset.

This confirms the period covered by the taxi trip records.

```python
print(
    "Minimum Pickup Date:",
    df['tpep_pickup_datetime'].min()
)

print(
    "Maximum Pickup Date:",
    df['tpep_pickup_datetime'].max()
)
```

# Step 13: Forecasting Feasibility Check

Verify that the required columns for hourly demand forecasting are available.

```python
required_cols = [
    'tpep_pickup_datetime',
    'PULocationID'
]

for col in required_cols:
    print(
        f"{col} Exists :",
        col in df.columns
    )
```
