### Key Observations

- **Yellow Taxi** provides a strong balance between dataset size, feature richness, and computational efficiency.
- **Green Taxi** contains very limited records and may not provide sufficient demand patterns for robust forecasting.
- **FHV** has a good number of trips but lacks several useful trip-level attributes such as fare amount, passenger count, and trip distance.
- **FHVHV** contains the highest number of trips and richest coverage but requires substantial computational resources and storage.

### Final Dataset Selection: Yellow Taxi

**Reason for Selection:**

1. Contains approximately **3.7 million trip records**, providing sufficient data for learning hourly demand patterns.
2. Includes important attributes such as:
   - Pickup and dropoff timestamps
   - Pickup and dropoff zones
   - Passenger count
   - Trip distance
   - Fare information
   - Payment information
3. Supports all required feature engineering steps:
   - Hourly aggregation
   - Day-of-week extraction
   - Weekend indicators
   - Lag features
   - Rolling statistics
4. Has a manageable memory footprint (**528 MB**) compared to FHVHV (**4.2 GB**).
5. Provides richer information than FHV while remaining computationally efficient.
6. Offers significantly more observations than Green Taxi, resulting in stronger and more reliable demand patterns.

### Conclusion

Although all four datasets are suitable for creating an hourly demand forecasting target, the **Yellow Taxi dataset** offers the best balance between data volume, feature availability, forecasting capability, and computational efficiency. Therefore, it was selected as the primary dataset for model development, training, and evaluation.
